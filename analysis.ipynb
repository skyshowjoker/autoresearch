# AutoQuant 实验分析
按 protocol 分组比较；final 不参与研发排名。score 越高越好。

```python
from pathlib import Path
import pandas as pd
import json
results = pd.read_csv('results.tsv', sep='\t')
results.tail(20)
```

```python
valid = results[(results.status == 'success') & (results.profile == 'dev')]
valid.sort_values(['protocol', 'score'], ascending=[True, False]).groupby('protocol').head(5)
```

```python
if not valid.empty:
    selected = Path(valid.iloc[-1].artifacts)
    display(pd.json_normalize(json.loads((selected / 'folds.json').read_text())))
    display(pd.read_parquet(next(selected.glob('*equity.parquet'))).tail())
```
